```python
'''Experiment 3: Bayesian Networks (LO2)
Task: Model conditional distributions over sequential graph hierarchies.
'''
import os
import pandas as pd

def perform_experiment_3(filename="bayesian_ready_data.csv"):
    if not os.path.exists(filename):
        print(f"Error: Target file '{filename}' not found. Run Script 1 first!")
        return

    # Step 1: Read the generated CSV dataset
    df = pd.read_csv(filename)
    print(f"--- Experiment 3: Bayesian Networks ({len(df):,} Rows Loaded) ---")

    # Step 2: Compute Root Node Prior Probability -> P(Rain = Yes)
    p_rain_yes = (df['Rain'] == 'Yes').mean()
    print(f"\n[Step 2: Root Prior]")
    print(f"P(Rain = Yes): {p_rain_yes:.4f}")

    # Step 3: Compute Conditional Probability Layer 1 -> P(HeavyTraffic | Rain)
    rain_yes_df = df[df['Rain'] == 'Yes']
    rain_no_df = df[df['Rain'] == 'No']

    p_traffic_given_rain = (rain_yes_df['HeavyTraffic'] == 'Yes').mean()
    p_traffic_given_clear = (rain_no_df['HeavyTraffic'] == 'Yes').mean()

    print(f"\n[Step 3: CPT Layer 1 - Traffic given Weather]")
    print(f"P(Heavy Traffic | Rain = Yes): {p_traffic_given_rain:.4f}")
    print(f"P(Heavy Traffic | Rain = No):  {p_traffic_given_clear:.4f}")

    # Step 4: Compute Conditional Probability Layer 2 -> P(Late | HeavyTraffic)
    traffic_yes_df = df[df['HeavyTraffic'] == 'Yes']
    traffic_no_df = df[df['HeavyTraffic'] == 'No']

    p_late_given_traffic = (traffic_yes_df['Late'] == 'Yes').mean()
    p_late_given_no_traffic = (traffic_no_df['Late'] == 'Yes').mean()

    print(f"\n[Step 4: CPT Layer 2 - Lateness given Traffic]")
    print(f"P(Late | Heavy Traffic = Yes): {p_late_given_traffic:.4f}")
    print(f"P(Late | Heavy Traffic = No):  {p_late_given_no_traffic:.4f}")

    # Step 5: Chain Rule Joint Probability Calculation
    theoretical_joint = p_rain_yes * p_traffic_given_rain * p_late_given_traffic
    empirical_joint = ((df['Rain'] == 'Yes') & (df['HeavyTraffic'] == 'Yes') & (df['Late'] == 'Yes')).mean()

    print(f"\n[Step 5: Network Chain Rule Verification]")
    print(f"Calculated Chain Rule Joint P(Rain, Traffic, Late): {theoretical_joint:.4f}")
    print(f"Direct Empirical Dataset Match Rate:                {empirical_joint:.4f}")
    print("-" * 65)

if __name__ == "__main__":
    perform_experiment_3()
```
