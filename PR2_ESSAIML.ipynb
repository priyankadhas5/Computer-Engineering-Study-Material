```python
'''
Experiment 2: Probability Concepts in AI (LO1)
Task: Roll dice simulations and utilize Bayes' Theorem.
Steps To be followed
1. Imports (numpy, random): Needed to generate millions of random rolls instantly and handle high-speed mathematical array operations.
2.Vectorized Simulation (randint): Simulates real-world randomness by generating unbiased, independent values for each die roll.
3.Summing Arrays: Combines the independent variables into a single compound variable (the total score) to study how multiple events interact.
4.Empirical Calculation: Computes the actual, real-world frequencies from your gathered experimental data.
5.Theoretical Calculation: Provides the perfect mathematical baseline so you can evaluate the accuracy of your AI simulation model.
6.Difference Calculation: Measures the variance to prove that massive datasets reduce real-world errors (Law of Large Numbers).
7.Bayes Prior Setup: Establishes your starting historical knowledge baseline before any new evidence alters your prediction model.
8.Total Evidence (P(B)): Acts as a normalizing factor, factoring in both true hits and false alarms to ensure the final probability adds up to 100%.
9.Posterior Computation: Updates the initial belief into a refined, data-backed prediction based on the newly observed evidence.
Course Teacher : Ms.Priyanka A.Hajare
'''
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# --- TASK 1: DICE ROLL SIMULATION ---
def simulate_dice(rolls=100000):
    print(f"--- Task 1: Dice Simulation ({rolls:,} rolls) ---")

    #1.1  Generate random outcomes for two six-sided dice
    die1 = np.random.randint(1, 7, size=rolls)
    die2 = np.random.randint(1, 7, size=rolls)
    sums = die1 + die2

    print(f"{'Sum':<6}{'Empirical %':<15}{'Theoretical %':<15}{'Diff %':<10}")
    print("-" * 50)

    #1.2  Calculate probabilities for all possible sums (2 to 12)
    for s in range(2, 13):
        empirical_prob = np.sum(sums == s) / rolls

        # Theoretical calculation: favorable combinations / 36 total outcomes.
        favorable_ways = 6 - abs(s - 7)
        theoretical_prob = favorable_ways / 36

        diff = abs(empirical_prob - theoretical_prob)
        print(f"{s:<6}{empirical_prob*100:<15.3f}{theoretical_prob*100:<15.3f}{diff*100:<10.3f}")
    print("\n")
  # 1.3  Graphical Sample Space Generation (You can skip this step as it is used to learn visualization of sample space)
    grid = [[d1 + d2 for d2 in range(1, 7)] for d1 in range(1, 7)]
    df = pd.DataFrame(
        grid,
        index=[f"Die 1: {i}" for i in range(1, 7)],
        columns=[f"Die 2: {j}" for j in range(1, 7)]
    )

    plt.figure(figsize=(8, 6))
    sns.heatmap(df, annot=True, fmt="d", cmap="coolwarm", linewidths=1, linecolor='white')
    plt.title("Graphical Table: Two Dice Sample Space Distribution", fontsize=12, pad=15)
    plt.tight_layout()
    plt.show()
# --- TASK 2: BAYES' THEOREM POSTERIOR UPDATER ---
def update_bayes_posterior(prior, sensitivity, specificity):
    """
    Computes posterior probability using Bayes' Theorem.
    Context: Diagnostic system evaluating if an event (A) occurred given evidence (B).
    Formula: P(A|B) = [P(B|A) * P(A)] / [P(B|A)*P(A) + P(B|not A)*P(not A)]
    """
    print("--- Task 2: Bayes' Theorem Update Diagnostic ---")

    p_event = prior                       # P(A)
    p_no_event = 1.0 - prior               # P(not A)

    p_evidence_given_event = sensitivity   # P(B|A)
    p_evidence_given_no_event = 1.0 - specificity # P(B|not A) - False Positive Rate

    # Marginal probability of evidence: P(B)
    p_evidence = (p_evidence_given_event * p_event) + (p_evidence_given_no_event * p_no_event)

    # Posterior probability: P(A|B)
    posterior = (p_evidence_given_event * p_event) / p_evidence

    print(f"Prior Probability P(Event):       {p_event:.4f}")
    print(f"Evidence Likelihood P(Ev|Event):  {p_evidence_given_event:.4f}")
    print(f"Total Evidence Probability P(Ev): {p_evidence:.4f}")
    print(f"Posterior Probability P(Event|Ev): {posterior:.4f}")

if __name__ == "__main__":
    # Execute simulations
    simulate_dice(rolls=100000)

    # Run diagnostic updater with sample inputs
    # E.g., Prior = 1%, Sensitivity = 99%, Specificity = 95%
    update_bayes_posterior(prior=0.01, sensitivity=0.99, specificity=0.95)
```
