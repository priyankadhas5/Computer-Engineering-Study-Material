```python
# PR5 Linear Regression.ipynb



#Data Loading and Analysis

import opendatasets as od
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Provide students with the exact link to YOUR dataset here
DATASET_URL = "https://www.kaggle.com/datasets/priyankahajare15/salarydataset"
# This will prompt students to type their Kaggle Username and API Key
od.download(DATASET_URL)
df = pd.read_csv("/content/salarydataset/salary_data.csv")
df.head()

#VISUALIZATION
plt.figure(figsize=(8,5))
plt.scatter(df["YearsExperience"],df["Salary"],color="red",label = "Data points")
plt.xlabel("Years of experience")
plt.ylabel("Salary")
plt.title("Year if Experience vs Salary")
plt.grid(True)
plt.legend()
plt.show()

#Prepare the Features (X) and Labels (y) Data Preprocessing
x = df[['YearsExperience']].values.astype(float)  # Input INDEPENDANT
y = df['Salary'].values.astype(float)             # Output DEPENDANT

#Model Implementation
from sklearn.linear_model import LinearRegression
#Train the Linear Regression Model.
model = LinearRegression()  #instance is created
model.fit(x.reshape(-1, 1), y)
#Display the Learned Parameters (Slope and Intercept)
print(f"Slope (m): {model.coef_[0]:.2f}")
print(f"Intercept (b): {model.intercept_:.2f}")

#Predict the Score for a Person Who has 7 Years of experience
predicted_score = model.predict([[7]])
print(f"Predicted Salary for 7 years of experience: {predicted_score[0]:,.2f}")

#Plot the Regression Line
plt.figure(figsize=(8, 5))
plt.scatter(x, y, color='blue', label="Data Points")
plt.plot(x, model.predict(x.reshape(-1, 1)), color='red', label="Regression Line")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear Regression Fit")
plt.grid(True)
plt.legend()
plt.show()

# Model Evaluation
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Generate predictions for the training data
y_pred = model.predict(x.reshape(-1, 1))

# Calculate Evaluation Metrics
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y, y_pred)

print("\n--- Model Evaluation ---")
print(f"Mean Absolute Error (MAE): {mae:,.2f}")
print(f"Mean Squared Error (MSE): {mse:,.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:,.2f}")
print(f"R-squared Score (R2): {r2:.4f}")
```
